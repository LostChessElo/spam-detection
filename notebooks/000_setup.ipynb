```python
# Imports
import random 
import kagglehub
import pandas as pd 
from pathlib import Path
```

```python
# download original dataset
SEED = 42
random.seed(SEED)

path = Path(kagglehub.dataset_download(
    "meruvulikith/190k-spam-ham-email-dataset-for-classification"
))

print("Path to the dataset: ", path)
```

```python
df = pd.read_csv(path / "spam_Emails_data.csv")
print(df.isna().sum())
df = df.dropna(subset=["text"])
spam = df[df["label"] == "Spam"].copy()
ham = df[df["label"] == "Ham"].copy()

spam = spam.sample(frac=1, random_state=SEED).reset_index(drop=True)
ham = ham.sample(frac=1, random_state=SEED).reset_index(drop=True)
```

```python
spam_split = int(len(spam) * 0.8)
ham_split = int(len(ham) * 0.8)

training_df = pd.concat([
    spam.iloc[:spam_split],
    ham.iloc[:ham_split]
])

testing_df = pd.concat([
    spam.iloc[spam_split:],
    ham.iloc[ham_split:]
])

training_df = training_df.sample(frac=1, random_state=SEED).reset_index(drop=True)
testing_df = testing_df.sample(frac=1, random_state=SEED).reset_index(drop=True)
```

```python
output_dir = Path("../data/raw")
output_dir.mkdir(parents=True, exist_ok=True)

training_df.to_csv(output_dir / "training_data.csv", index=False)
testing_df.to_csv(output_dir / "testing_dataset.csv", index=False)
```
