```python
from pathlib import Path
import urllib.request

release_url = "https://github.com/LostChessElo/spam-detection/releases/download/v1.0.0/mnb-spam-detection-v1.0.0.zip"
zip_path = Path("/content/mnb-spam-detection-v1.0.0.zip")

urllib.request.urlretrieve(release_url, zip_path)

print(f"Project downloaded to: {zip_path}")
```

```python
import zipfile
from pathlib import Path

project_dir = Path("/content/mnb-spam-detection-v1.0.0")

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("/content")

print(f"Project extracted to: {project_dir}")
print("Project is ready.")
```
